```python
df["Stockout_Risk"] = df["Inventory Level"] < df["Units Sold"]

stockout_products = df[df["Stockout_Risk"]]

stockout_products.shape
```

```python
df["Overstock_Risk"] = df["Inventory Level"] > (df["Units Sold"] * 2)

overstock_products = df[df["Overstock_Risk"]]

overstock_products.shape
```

```python
print("Stockout Risk Records:", stockout_products.shape[0])
print("Overstock Risk Records:", overstock_products.shape[0])
```

```python
df["Inventory_Gap"] = df["Inventory Level"] - df["Units Sold"]

df[["Product ID", "Inventory Level", "Units Sold", "Inventory_Gap"]].head()
```

```python
def classify_inventory(row):
    if row["Inventory_Gap"] < 0:
        return "UNDERSTOCKED"
    elif row["Inventory_Gap"] > 100:
        return "OVERSTOCKED"
    else:
        return "OPTIMAL"

df["Inventory_Status"] = df.apply(classify_inventory, axis=1)

df["Inventory_Status"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["Inventory_Status"].value_counts().plot(kind="bar")
plt.title("Inventory Health Status Distribution")
plt.show()
```
